```python
import pandas as pd
import numpy as np
```

```python
# Load Training data
df = pd.read_csv(R'C:\Users\veere\OneDrive\Desktop\projectD12\DiseasePredictionUsingMachineLearning\Notebook\Training.csv')
df.head()
```

```python
print(df)

```

```python
X = df.iloc[:, :-1]
y = df['prognosis']

# Train, Test split
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=20)
```

```python
# Load real test data
df_test = pd.read_csv(r'C:\Users\veere\OneDrive\Desktop\projectD12\DiseasePredictionUsingMachineLearning\Notebook\Testing.csv')
X_acutal_test = df_test.iloc[:, :-1]
y_actual_test = df_test['prognosis']
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
knn_classifier= KNeighborsClassifier(n_neighbors=5, metric='minkowski', p=2 )  
knn_classifier.fit(X_train, y_train)  
y_pred= knn_classifier.predict(X_test)
print(accuracy_score(y_test, y_pred))
print("Accuracy on acutal test: ", knn_classifier.score(X_acutal_test, y_actual_test))
result_knn=knn_classifier.score(X_acutal_test, y_actual_test)
```

```python
from sklearn.metrics import accuracy_score
from sklearn.svm import SVC # "Support vector classifier"  
svc_classifier = SVC(kernel='linear', random_state=0)  
svc_classifier.fit(X_train, y_train)
y_pred= svc_classifier.predict(X_test)
print(accuracy_score(y_test, y_pred))
print("Accuracy on acutal test: ", svc_classifier.score(X_acutal_test, y_actual_test))
result_svm=svc_classifier.score(X_acutal_test, y_actual_test)
```

```python
from sklearn.metrics import accuracy_score

from sklearn.linear_model import LogisticRegression  
lr_classifier= LogisticRegression(random_state=0) 

lr_classifier.fit(X_train, y_train)
y_pred= lr_classifier.predict(X_test)
print(accuracy_score(y_test, y_pred))
print("Accuracy on acutal test: ", lr_classifier.score(X_acutal_test, y_actual_test))
result_lr=lr_classifier.score(X_acutal_test, y_actual_test)
```

```python
from sklearn.metrics import accuracy_score

from sklearn.naive_bayes import GaussianNB  
nb_classifier = GaussianNB()

nb_classifier.fit(X_train, y_train)
y_pred= nb_classifier.predict(X_test)
print(accuracy_score(y_test, y_pred))
print("Accuracy on acutal test: ", nb_classifier.score(X_acutal_test, y_actual_test))
result_db=(nb_classifier.score(X_acutal_test, y_actual_test))
```

```python
from sklearn.metrics import accuracy_score

from sklearn.tree import DecisionTreeClassifier  
dt_classifier= DecisionTreeClassifier(criterion='entropy', random_state=0) 

dt_classifier.fit(X_train, y_train)
y_pred= dt_classifier.predict(X_test)
print(accuracy_score(y_test, y_pred))
print("Accuracy on acutal test: ", dt_classifier.score(X_acutal_test, y_actual_test))
result_DT=dt_classifier.score(X_acutal_test, y_actual_test)
result_DT
```

```python
from sklearn.metrics import accuracy_score

# Random Forest Classifier
from sklearn.ensemble import RandomForestClassifier
rf_classifier = RandomForestClassifier()

rf_classifier.fit(X_train, y_train)
y_pred= rf_classifier.predict(X_test)
print(accuracy_score(y_test, y_pred))
print("Accuracy on acutal test: ", rf_classifier.score(X_acutal_test, y_actual_test))
result_RF=rf_classifier.score(X_acutal_test, y_actual_test)
```

```python
symptoms_dict = {}
for index, symptom in enumerate(X):
    symptoms_dict[symptom] = index
print(symptoms_dict,end="\n\n")

input_vector = np.zeros(len(symptoms_dict))
input_vector[[symptoms_dict['itching'], symptoms_dict['skin_rash']]] = 1
print(input_vector,end="\n\n")


print(rf_classifier.predict_proba([input_vector]),end="\n\n")
print(rf_classifier.predict([input_vector]),end="\n\n")
```
